This Script helps me to count the data points of each type of road

```python
# Import libraries
import os
import csv
```

```python
def count_csv_files(folder_path):
    if not os.path.isdir(folder_path):
        raise ValueError(f"The provided path '{folder_path}' is not a valid directory.")
    
    csv_files = [f for f in os.listdir(folder_path) if f.lower().endswith('.csv')]

    return len(csv_files)
```

```python
def find_dir(folder_path):
    if not os.path.isdir(folder_path):
        raise ValueError(f"The provided path '{folder_path}' is not a valid directory.")
    
    folder_list = [f for f in os.listdir(folder_path)]

    return folder_list
```

```python
def location_csv_generator(road_type, Data_of, location):
    os.makedirs(location, exist_ok=True)
    os.makedirs(f"{location}/{Data_of}", exist_ok=True)
    count_of_datapoints = 0
    with open(f'{location}/{Data_of}/{road_type}.csv', mode='w', newline='') as file:
        writer = csv.writer(file)
        writer.writerow(['Directory Name', 'Number of CSV Files'])
        data_dir = f"Harshit_Preprocessed_Data-{Data_of}"
        sub_dir = find_dir(os.path.join(data_dir, road_type))
        for sub_sub_dir in sub_dir:
            path = os.path.join(data_dir, road_type, sub_sub_dir)
            try:
                num_csv_files = count_csv_files(path)
                # print(f"Number of CSV files in '{path}': {num_csv_files}")
                writer.writerow([path, num_csv_files])
                count_of_datapoints += num_csv_files
            except ValueError as e:
                print(e)
    print(f"Total number of datapoints in '{road_type}' for '{Data_of}': {count_of_datapoints}")
    return count_of_datapoints
```

```python
road_types = ['Bitumin', 'Block', 'Concrete', 'Kanker']
chunk_sizes = ['CS200', 'CS250', 'CS300']
```

```python
for road_type in road_types:
    for chunk_size in chunk_sizes:
        location_csv_generator(road_type=road_type, Data_of=chunk_size,location="CounterOutputFiles")
```
