```python
# Import libraries
import os
import csv
```

```python
def count_csv_files(folder_path):
    if not os.path.isdir(folder_path):
        raise ValueError(f"The provided path '{folder_path}' is not a valid directory.")
    
    csv_files = [f for f in os.listdir(folder_path) if f.lower().endswith('.csv')]

    return len(csv_files)
```

```python
def find_dir(folder_path):
    if not os.path.isdir(folder_path):
        raise ValueError(f"The provided path '{folder_path}' is not a valid directory.")
    
    folder_list = [f for f in os.listdir(folder_path)]

    return folder_list
```

```python
find_dir("Harshit_Preprocessed_Data-CS300/Bitumin")
```

```python
count_csv_files("Harshit_Preprocessed_Data-CS300/Bitumin/001_Abj_Bitumin_1_12-2")
```

```python
for dir in find_dir("Harshit_Preprocessed_Data-CS300/Bitumin"):
    print(count_csv_files(f"Harshit_Preprocessed_Data-CS300/Bitumin/{dir}"))
```
